```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/placement.csv') # Function To read CSV files and pandas function is stored in df
```

```python
df.head() # when we have to take a overview of pandas Dataframe
```

```python
df.shape
```

```python
df.info() #it gives the info of every column
```

```python
df = df.iloc[:,1:]
```

```python
df.head() # It just remove the 0th index value or it just remove column "unnamed"
```

```python
#2 EDA it is used to plot the graphs to analyze the data
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['cgpa'],df['iq']) # means we plot a graph in which 'cgpa' is on x axis and 'iq' is on y axis
```

```python
plt.scatter(df['cgpa'],df['iq'],c=df['placement']) # to find whose placement are done we just add placement column on this
```

```python
# Extract the Input and Output
```

```python
X = df.iloc[:,0:2] # all rows but first 2 columns
Y = df.iloc[:,-1] # all rows but last column
```

```python
X
```

```python
X.shape
```

```python
Y
```

```python
Y.shape
```

```python
#6 Train test split
```

```python
from sklearn.model_selection import train_test_split #importing a library
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.1) # it means 10%(rows) will be go for test data and the remaining 90%(rows) will go fo training
```

```python
X_train # random data will be go in the training data for X(Input)
```

```python
Y_train # random data will be picked for Y(output)
```

```python
#5 scaling the data
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train) # we change the data in the form of array
```

```python
X_train
```

```python
X_test = scaler.transform(X_test)
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression # the logistic regression class will be imported froom this
```

```python
clf = LogisticRegression() # we make a variable that stores the logistic regression class
```

```python
# Model Training
clf.fit(X_train,Y_train)    # it has a power to access the classes of logistic regression
```

```python
#8 Evaluate The model / In this we find the accuracy for test data
```

```python
clf.predict(X_test) #input values of left data which was not used for training
```

```python
Y_test # as we see the model was predicted some wrong values
```

```python
#Now we compare side by side how many prediction were wrong and how many was right
```

```python
Y_pred = clf.predict(X_test) # store those values in the variable
# but the shape of both Y_test and Y_pred was equal
```

```python
from sklearn.metrics import accuracy_score # it is the matrics function
```

```python
accuracy_score(Y_test,Y_pred) # it calculates the accuracy automatically
# it means our accuracy is 80%
```

```python
# what patterns should be noticed by our ML model that can be visualise by the decision boundary
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(X=X_train,Y=Y_train.values,clf=clf,legend=2)
# Y_train.values because the Y is not the form of arrary that's why we use .values
# .values is used to convert the data into numpy array
# and our classifier is clf so clf=clf
# it gives the decision boundary for the model
```
